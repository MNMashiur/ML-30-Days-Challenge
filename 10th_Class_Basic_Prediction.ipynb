**Importing Libraries**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/content/car_purchasing.csv', encoding='latin1')

df.head()
```

```python
for col in df.columns:
    print(repr(col))
```

```python
df.columns = df.columns.str.strip()
```

```python
df.drop(
    columns=[
        'customer name',
        'customer e-mail',
        'country',
        'credit card debt',
        'net worth',
        'gender'
    ],
    inplace=True
)
```

```python
df.head()
```

```python
df.rename(
    columns={
        'age': 'Age',
        'annual Salary': 'Salary',
        'car purchase amount': 'Buy'
    },
    inplace=True
)

df
```

```python
df['Buy'] = (df['Buy'] > df['Buy'].median()).astype(int)
```

**Segegate Dataset into X, Y**

```python
x= df.iloc[:,:-1]

x.head()
```

```python
y= df.iloc[:, -1]

y.head()
```

**Split Dataset**

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test= train_test_split(x,y, test_size=0.25, random_state=0)
```

**Feature Scalling**

```python
from sklearn.preprocessing import StandardScaler

sc= StandardScaler()
x_train= sc.fit_transform(x_train)
x_test= sc.transform(x_test)
```

**Model Training**

```python
from sklearn.linear_model import LogisticRegression

model= LogisticRegression(random_state=0)
model.fit(x_train, y_train)
```

**Prediction for new Customer**

```python
age= int(input("Enter New Customers Age: "))
salary= int(input("Enter New Customers Salary: "))
newCust= [[age, salary]]
result= model.predict(sc.transform(newCust))
print(result)

if result==1:
  print("Customer will Buy")
else:
  print("Customer will not Buy")
```

**Prediction for all test data**

```python
y_pred= model.predict(x_test)
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score

cm= confusion_matrix(y_test, y_pred)
print("Confusion Matrix: ")
print(cm)

print('Accuracy of the model: {0}%'. format(accuracy_score(y_test, y_pred)*100))
```
